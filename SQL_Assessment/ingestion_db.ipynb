```python
import pandas as pd
import os
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
```

```python
url = URL.create(
    drivername="mysql+mysqldb",
    username="root",
    password=os.getenv("MYSQL_PASSWORD"),
    host="127.0.0.1",
    port=3306,
    database="section_b"
)

engine = create_engine(url)
```

```python
df = pd.read_csv(r"C:\Users\allex\Downloads\Sample - Superstore.csv", encoding="latin1")
df.head()
```

```python
df = df[['Order ID','Order Date','Sales','Quantity','Discount','Profit']]
df.head()
```

```python
df.info()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df.info()
```

```python
df.to_sql('orders',con=engine,if_exists='replace',index= False, method='multi', chunksize=5000)
```

```python
# Export to CSV without index column
df.to_csv('orders.csv', index=False)
```
