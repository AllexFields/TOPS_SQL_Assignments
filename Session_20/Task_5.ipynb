```python
import pandas as pd
import os
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
```

```python
ipl_data = pd.DataFrame(pd.read_csv(r"C:\Users\allex\Downloads\IPL2025Batters.csv"))
```

```python
ipl_data.head()
```

```python
mi_batters = ipl_data[ipl_data['Team'] == 'MI']
mi_batters.head()
```

```python
# Establishing connection....

url = URL.create(
    drivername="mysql+mysqldb",
    username="root",
    password=os.getenv("MYSQL_PASSWORD"),
    host="127.0.0.1",
    port=3306,
    database="session_20"
)

engine = create_engine(url)
```

```python
try:
    mi_batters.to_sql('mi_batters', con=engine, if_exists= 'append', index= False, method='multi')

except Exception as e:
    print(e)
```
